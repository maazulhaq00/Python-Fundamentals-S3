```python
from abc import ABC, abstractmethod

# Abstract Base Class
```

```python
# Abstract Class - Vehicle
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass    
    @abstractmethod
    def stop(self):
        pass
    def info(self):
        print("This is a vehicle")
```

```python
# Concrete Class - Bike
class Bike(Vehicle):
    def start(self):
        print("Bike started with a kick")    
    def stop(self):
        print("Bike stoped")
```

```python
# Concrete Class - Car

class Car(Vehicle):
    def start(self):
        print("Car engine started")    
    def stop(self):
        print("Car stoped")
```

```python
b1 = Bike()

b1.start()
b1.stop()
b1.info()
```

```python
c1 = Car()

c1.start()
c1.stop()
c1.info()
```
