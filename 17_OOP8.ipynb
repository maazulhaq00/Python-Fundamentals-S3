```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    
    def deposit(self, amount):
        if(amount>0):
            self.__balance += amount
        else: 
            print("Invalid deposit")
    
    def show_balance(self):
        print(f"The balance is {self.__balance}")
        
```

```python
acc2 = BankAccount(2345)
acc2.show_balance()
```

```python
acc2.deposit(-1300)
```

```python
acc2.deposit(1300)
```
